```python
import numpy as np
import matplotlib.pyplot as plt

# Input range
x = np.linspace(-10, 10, 100)

# Weight and bias
w = 1.2
b = -2.0

# Linear combination
z = w * x + b

# Sigmoid activation
y= 1 / (1 + np.exp(-z))

# Plot
plt.plot(x, y, label="Sigmoid Output", color='purple')
plt.title("Neuron Output with Sigmoid Activation")
plt.xlabel("Input (x)")
plt.ylabel("Output (y)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Input range
x = np.linspace(-10, 10, 100)

# Weight and bias
w = 1.0
b = 0.0

# Linear combination
z = w * x + b

# Step activation (Heaviside function)
y = np.where(z >= 0, 1, 0)

# Plot
plt.plot(x, y, label="Step Activation", color='red', linewidth=2)
plt.title("Neuron Output with Step Activation")
plt.xlabel("Input (x)")
plt.ylabel("Output (y)")
plt.ylim(-0.1, 1.1)  # to make 0 and 1 clear
plt.grid(True)
plt.legend()
plt.show()
```
